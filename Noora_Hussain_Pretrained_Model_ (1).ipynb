1. Task A — Sentiment Classification
- Problem: Classify restaurant reviews by sentiment into three categorical classes.

- Target Audience / User: Restaurant owners and managers looking to monitor customer satisfaction.

- Input: Raw customer review text (in Arabic and English).

- Output: Predicted sentiment label — Positive, Neutral, or Negative.

- Evaluation & Success Metrics: Compare predicted labels against the manually created true labels (true_label) using Accuracy, Precision, Recall, F1-Score, and a Confusion Matrix.

2. Task B — Zero-Shot Topic Classification
- Problem: Automatically identify the primary topic or aspect discussed within a customer review without requiring task-specific training data.

- Target Audience / User: Restaurant operations teams seeking granular feedback on specific business pillars.

- Input: Restaurant review text.

- Output: Categorized topic label selected from predefined operational domains: Food Quality, Service, Price, Cleanliness, Atmosphere, Location, or Waiting Time.

- Evaluation & Success Metrics: Compare zero-shot model predictions against manually assigned ground-truth topic labels.

Task A Dataset :

- Size: 51 annotated restaurant reviews (reflecting a balanced distribution across sentiment classes).

Class Distribution:

- Positive: 23 reviews

- Negative: 16 reviews

- Neutral: 12 reviews

- Language Coverage: Bilingual dataset featuring reviews in both Arabic (33 reviews) and English (18 reviews).

Task B Dataset :

- Size: 35 manually labeled evaluation reviews.

- Target Aspects: Food Quality, Service, Price, Cleanliness, Atmosphere, Location, and Waiting Time.

```python
def add_metadata(prediction):
    return {"label": prediction["label"],
        "score": prediction["score"],
        "metadata": "huggingface_AI_model"}
```

# Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

```python
def Evaluate_Model(true_labels, predicted_labels):
    print("Accuracy: ", accuracy_score(true_labels, predicted_labels))
    print("Precision: ", precision_score(true_labels, predicted_labels, average="weighted", zero_division=0))
    print("Recall: ", recall_score(true_labels, predicted_labels, average="weighted", zero_division=0))
    print("F1 Score: ", f1_score(true_labels, predicted_labels, average="weighted", zero_division=0))
    print()
    print(classification_report(true_labels, predicted_labels, zero_division=0))
```

# Task B — Zero-Shot Classification

```python
classifier = pipeline(
    "zero-shot-classification",
    model="MoritzLaurer/mDeBERTa-v3-base-mnli-xnli"
)
```

```python
Sequence_classify = "The prices are a bit high for the portion sizes we received."
candidate_labels = ["Food Quality","Service","Price","Cleanliness","Atmosphere","Location","Waiting Time"]

classifier(
    Sequence_classify,
    candidate_labels,
    multi_label=False
)
```

```python
Zero_example = classifier(
    Sequence_classify,
    candidate_labels,
    multi_label=False)

Zero_output = {
    "label": Zero_example["labels"][0],
    "score": Zero_example["scores"][0],
    "metadata": "huggingface_AI_model"}

Zero_output
```

```python
Zero_predictions = []
Zero_scores = []

for review in T2_df["review_text"]:
    result = classifier(
        review,
        candidate_labels,
        multi_label=False
      )

    Zero_predictions.append(result["labels"][0])
    Zero_scores.append(result["scores"][0])
```

```python
T2_df["predicted_topic"] = Zero_predictions
T2_df["score"] = Zero_scores

T2_df.head()
```

```python
print("Accuracy:", accuracy_score(
    T2_df["topic_label"],
    T2_df["predicted_topic"]))

print("F1 Score:", f1_score(
    T2_df["topic_label"],
    T2_df["predicted_topic"],
    average="weighted",
    zero_division=0))
```

```python
print(classification_report(
    T2_df["topic_label"],
    T2_df["predicted_topic"],
    zero_division=0))
```

```python
T2_df["correct"] = (
    T2_df["topic_label"] == T2_df["predicted_topic"])

zero_errors = T2_df[
    T2_df["correct"] == False]

[[
    "review",
    "true_topic",
    "predicted_topic",
    "score"]]
zero_errors.head(10)
```
